```python
import pandas as pd
import numpy as np

# 1. Read and Prepare Data

input_filename = 'C:/Users/hp/Downloads/labs.csv'

# Define column names
test_column_name = 'Component_name'
csn_column_name = 'CSN'
value_column_name = 'Component_value'


df = pd.read_csv(input_filename)


df[test_column_name] = df[test_column_name].str.strip()

# Enhanced Cleaning: Extract numeric values from strings like '<0.012'
df[value_column_name] = df[value_column_name].astype(str).str.extract(r'(\d+\.?\d*)', expand=False)

# Ensure the value column is numeric; non-numeric values become NaN (null)
df[value_column_name] = pd.to_numeric(df[value_column_name], errors='coerce')

# Drop rows that do not have a valid numeric result
df.dropna(subset=[value_column_name], inplace=True)


# 2. Group similar Component_name

df['normalized_test'] = df[test_column_name]


pco2_arterial_list = ['PCO2 (A)', 'POC:PCO2 (A), ISTAT', 'PCO2.ARTERIAL']
df.loc[df[test_column_name].isin(pco2_arterial_list), 'normalized_test'] = 'PCO2, Arterial (Combined)'
pco2_venous_list = ['PCO2.VENOUS', 'POC:PCO2 (V), ISTAT']
df.loc[df[test_column_name].isin(pco2_venous_list), 'normalized_test'] = 'PCO2, Venous (Combined)'
ph_arterial_list = ['PH.ARTERIAL', 'POC:PH (A), ISTAT']
df.loc[df[test_column_name].isin(ph_arterial_list), 'normalized_test'] = 'pH, Arterial (Combined)'
ph_venous_list = ['PH.VENOUS', 'POC:PH (V), ISTAT']
df.loc[df[test_column_name].isin(ph_venous_list), 'normalized_test'] = 'pH, Venous (Combined)'
po2_arterial_list = ['PO2 (A)', 'POC:PO2 (A), ISTAT', 'PO2.ARTERIAL']
df.loc[df[test_column_name].isin(po2_arterial_list), 'normalized_test'] = 'PO2, Arterial (Combined)'
po2_venous_list = ['PO2.VENOUS', 'POC:PO2 (V), ISTAT']
df.loc[df[test_column_name].isin(po2_venous_list), 'normalized_test'] = 'PO2, Venous (Combined)'
blood_type_regex = r'abo/rh|blood type verification'
df.loc[df[test_column_name].str.contains(blood_type_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Blood Typing & RH (Combined)'
mask_antibody_screen = df[test_column_name].str.contains('ANTIBODY SCREEN', case=False, na=False)
df.loc[mask_antibody_screen, 'normalized_test'] = 'Antibody Screen (Combined)'
basophil_pct_regex = r'BASOPHILS? \s* % \s* \( (AUTO|MANUAL) DIFF \)'
df.loc[df[test_column_name].str.contains(basophil_pct_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Basophil % (Combined)'
mask_basophil_abs = df[test_column_name].str.contains(r'BASOPHILS?.*ABSOLUTE', case=False, na=False, regex=True)
df.loc[mask_basophil_abs, 'normalized_test'] = 'Basophil Absolute (Combined)'
mask_basophil_csf = df[test_column_name].str.contains(r'BASOPHILS?.*CSF', case=False, na=False, regex=True)
df.loc[mask_basophil_csf, 'normalized_test'] = 'Basophil CSF (Combined)'
mask_blasts = df[test_column_name].str.contains('BLASTS', case=False, na=False)
df.loc[mask_blasts, 'normalized_test'] = 'Blasts (Combined)'
eos_pct_regex = r'EOSINOPHILS? \s* % \s* \( (AUTO|MANUAL) DIFF \)'
df.loc[df[test_column_name].str.contains(eos_pct_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Eosinophil % (Combined)'
mask_eos_abs = df[test_column_name].str.contains(r'EOSINOPHILS?.*ABSOLUTE', case=False, na=False, regex=True)
df.loc[mask_eos_abs, 'normalized_test'] = 'Eosinophil Absolute (Combined)'
mask_esr = df[test_column_name].str.contains('ESR', case=False, na=False)
df.loc[mask_esr, 'normalized_test'] = 'ESR (Combined)'
hct_hgb_regex = r'^HCT|HEMOGLOBIN'
df.loc[df[test_column_name].str.contains(hct_hgb_regex, case=False, na=False, regex=True), 'normalized_test'] = 'HCT & Hemoglobin (Combined)'
mono_pct_regex = r'MONOCYTES? \s* % \s* \( (AUTO|MANUAL) DIFF \)'
df.loc[df[test_column_name].str.contains(mono_pct_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Monocyte % (Combined)'
mono_abs_regex = r'MONOCYTE.*ABSOLUTE'
df.loc[df[test_column_name].str.contains(mono_abs_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Monocyte Absolute (Combined)'
rbc_count_regex = r'NORMAL RBC ON .* COUNT'
df.loc[df[test_column_name].str.contains(rbc_count_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Normal RBC Count (Combined)'
mask_nrbc = df[test_column_name].str.contains('NRBC', case=False, na=False)
df.loc[mask_nrbc, 'normalized_test'] = 'NRBC (Combined)'
rbc_small_square_regex = r"RBC'S IN SMALL SQUARE"
df.loc[df[test_column_name].str.contains(rbc_small_square_regex, case=False, na=False, regex=True), 'normalized_test'] = "RBC's in Small Square (Combined)"
lymph_abs_regex = r'LYMPHOCYTES?.*ABSOLUTE'
df.loc[df[test_column_name].str.contains(lymph_abs_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Lymphocyte Absolute (Combined)'
mask_adeno = df[test_column_name].str.contains('ADENOVIRUS', case=False, na=False)
df.loc[mask_adeno, 'normalized_test'] = 'Adenovirus (Combined)'
mask_hav = df[test_column_name].str.contains('ANTI-HAV IGG', case=False, na=False)
df.loc[mask_hav, 'normalized_test'] = 'Anti-HAV IgG (Combined)'
mask_hbc = df[test_column_name].str.contains('ANTI-HBC', case=False, na=False)
df.loc[mask_hbc, 'normalized_test'] = 'Anti-HBc (Combined)'
mask_hcv = df[test_column_name].str.contains('ANTI-HCV', case=False, na=False)
df.loc[mask_hcv, 'normalized_test'] = 'Anti-HCV (Combined)'
mask_ecoli = df[test_column_name].str.contains('ESCHERICHIA COLI', case=False, na=False)
df.loc[mask_ecoli, 'normalized_test'] = 'Escherichia Coli (Combined)'
hbsag_regex = r'HBSAG|HEP B SURFACE'
df.loc[df[test_column_name].str.contains(hbsag_regex, case=False, na=False, regex=True), 'normalized_test'] = 'Hepatitis B Surface Antigen (Combined)'
mask_rhinovirus = df[test_column_name].str.contains('RHINOVIRUS', case=False, na=False)
df.loc[mask_rhinovirus, 'normalized_test'] = 'Rhinovirus (Combined)'
mask_bili_unconj = df[test_column_name].str.contains('BILIRUBIN, UNCONJ', case=False, na=False)
df.loc[mask_bili_unconj, 'normalized_test'] = 'Bilirubin, Unconjugated (Combined)'
mask_calcium = df[test_column_name].str.contains('CALCIUM', case=False, na=False)
df.loc[mask_calcium, 'normalized_test'] = 'Calcium (Combined)'
mask_hdl = df[test_column_name].str.contains('CHOLESTEROL.*HDL', case=False, na=False, regex=True)
df.loc[mask_hdl, 'normalized_test'] = 'Cholesterol, HDL (Combined)'
mask_egfr = df[test_column_name].str.contains('EGFR', case=False, na=False)
df.loc[mask_egfr, 'normalized_test'] = 'eGFR (Combined)'
mask_ldl = df[test_column_name].str.contains('LDL', case=False, na=False)
df.loc[mask_ldl, 'normalized_test'] = 'LDL (Combined)'
potassium_list = ['POTASSIUM', 'POC:POTASSIUM, ISTAT', 'POTASSIUM.WHOLE BLD']
df.loc[df[test_column_name].isin(potassium_list), 'normalized_test'] = 'Potassium (Combined)'
sodium_list = ['SODIUM', 'POC:SODIUM, ISTAT', 'SODIUM.WHOLE BLD']
df.loc[df[test_column_name].isin(sodium_list), 'normalized_test'] = 'Sodium (Combined)'

print("Component_name has been grouped.")

#3. Counts the frequency

test_counts = df['normalized_test'].value_counts()
high_frequency_tests = test_counts[test_counts > 10000].index.tolist()
print(f"Found {len(high_frequency_tests)} tests meeting the criteria.")
print(high_frequency_tests)

#4. Keep only the high-frequency
df_filtered = df[df['normalized_test'].isin(high_frequency_tests)].copy()

#5. Aggregate data: Group by CSN and test name, then calculate the mean
aggregated_long_df = df_filtered.groupby([csn_column_name, 'normalized_test'])[value_column_name].mean().reset_index()

final_wide_df = aggregated_long_df.pivot(
    index=csn_column_name,
    columns='normalized_test',
    values=value_column_name
).reset_index()
final_wide_df.columns.name = None

#7. Save the new lab file
output_filename = 'labs_grouped.csv'
final_wide_df.to_csv(output_filename, index=False, encoding='utf-8-sig')

print(final_wide_df.head())
```
